```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('../Dataset/UCRArchive_2018/UMD/UMD_TEST.tsv', sep='\t')
```

```python
df
```

```python
dados_serie = df.iloc[0, 1:].values
```

```python
dados_serie
```

```python
fig = plt.figure(figsize=(6, 4), dpi=100)
ax = fig.add_subplot(111)

# Plotando os dados
ax.plot(dados_serie, marker="o", linestyle="-", color="#2c3e50")
ax.set_title("Série Temporal")
ax.set_xlabel("Tempo")
ax.set_ylabel("Valor")
plt.show()
```
