```python
import pandas as pd
from matplotlib.figure import Figure
import matplotlib.pyplot as plt
import numpy as np
from app.model.DynamicTimeWarping import DynamicTimeWarping, Distancia
import dtaidistance
```

```python
test_df = pd.read_csv('../Dataset/UCRArchive_2018/UMD/UMD_TEST.tsv', sep='\t')
train_df = pd.read_csv('../Dataset/UCRArchive_2018/UMD/UMD_TRAIN.tsv', sep='\t')
```

```python
dados_teste = test_df.iloc[0, 1:].values
dados_treino = train_df.iloc[0, 1:].values
```

```python
dados_teste
```

```python
dados_treino
```

```python
distancia, matriz_custo = dtaidistance.dtw.warping_paths(dados_teste, dados_treino)
print(distancia)
print(matriz_custo)
```

```python
dynamic_time_warping = DynamicTimeWarping(janela_de_busca=0, distancia=Distancia.EUCLIDIANA)
distancia = dynamic_time_warping.dtw_distance(dados_teste, dados_treino)
matriz_custo = dynamic_time_warping.gerar_matriz_dtw(dados_teste, dados_treino)
print(distancia)
print(matriz_custo)
```

```python
melhor_caminho = dtaidistance.dtw.warping_path(dados_teste, dados_treino)
print(melhor_caminho)
```

```python
melhor_caminho = dynamic_time_warping.dtw_warping_path(dados_teste, dados_treino)
print(melhor_caminho)
```

```python
fig = plt.figure(figsize=(8, 8))
gs = fig.add_gridspec(2, 2, width_ratios=[1, 4], height_ratios=[1, 4],
                      wspace=0.05, hspace=0.05)

# Subplot 1: Série Superior (Eixo X)
ax_x = fig.add_subplot(gs[0, 1])
ax_x.plot(dados_treino, color='g')
ax_x.set_axis_off()

# Subplot 2: Série Lateral (Eixo Y)
ax_y = fig.add_subplot(gs[1, 0])
ax_y.plot(dados_teste, np.arange(len(dados_teste)), color='r')
ax_y.invert_xaxis()  # Inverter para ficar de frente para a matriz
ax_y.set_axis_off()

# Subplot 3: A Matriz de Alinhamento
ax_matriz = fig.add_subplot(gs[1, 1])
ax_matriz.imshow(matriz_custo, interpolation='nearest', cmap='viridis', origin='lower')

# Plotar o caminho de alinhamento (a linha branca/preta na diagonal)
path_x, path_y = zip(*melhor_caminho)
ax_matriz.plot(path_y, path_x, color='white', linewidth=2)

ax_matriz.set_xlabel("Série B")
ax_matriz.set_ylabel("Série A")

plt.show()
```
